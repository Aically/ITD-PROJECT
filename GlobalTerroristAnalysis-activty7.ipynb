<h1>Regional Incident Distribution</h1>
<h1>visualize where incidents are concentrated in South Asia since 2000.

```python
# Filter for South Asia region and incidents after 2000
region_df = df[(df['region_txt'] == 'South Asia') & (df['iyear'] >= 2000)]

# Group by location
region_counts = region_df.groupby(['longitude', 'latitude']).size().reset_index(name='incident_count')

# Prepare heatmap data
heat_df = region_counts[['latitude', 'longitude', 'incident_count']].values.tolist()

# Create heatmap
my_map = folium.Map(location=[25, 80], zoom_start=5)
HeatMap(heat_df).add_to(my_map)
my_map
```

<h1>where Armed Assault attacks are concentrated globally.

<h1>Attack Type Concentration </h1>

```python
# Filter for 'Armed Assault' attacks
attack_df = df[df['attacktype1_txt'] == 'Armed Assault']

# Group by location
attack_counts = attack_df.groupby(['longitude', 'latitude']).size().reset_index(name='incident_count')

# Prepare heatmap data
heat_df = attack_counts[['latitude', 'longitude', 'incident_count']].values.tolist()

# Create heatmap
my_map = folium.Map(location=[20, 0], zoom_start=2)
HeatMap(heat_df).add_to(my_map)
my_map
```
